```python
# useful string methods

capitalize() # Converts the first character to upper case. for example if forgotten or converting sentences.
count() # Counts the number of times a specified value occurs in a string
lower() # Converts string into lower case. Python is case sensitive so it can be useful to convert the whole string into lower case.
upper()	# Converts a string into upper case.
swapcase() # Swaps cases.
splitlines() # Splits the string at line breaks and returns a list.
```

```python
# GC content

dna = "CCACCCTCGTGGTATGGCTAGGCATTCAGGAACCGGAGAACGCTTCAGACCAGCCCGGACTGGGAACCTGCGGGCAGTAGGTGGAAT"
count = dna.count('G') + dna.count('C')
dna_lenght = len(dna)
GC_content = (count / dna_lenght)*100
print(round(GC_content), '%')
```

```python
# the final value of radiance

radiance = 1.0
radiance = max(2.1, 2.0 + min(radiance, 1.1 * radiance - 0.5))
print (radiance)
```

```python
easy_string = "abc"

# prints the last letter of the string 'c'
print(max(easy_string))

rich = "gold"
poor = "tin"

# prints the later variable of the two, in this case 'tin'
print(max(rich, poor))

# prints the lenght of the longer word '4'
print(max(len(rich), len(poor)))
```
