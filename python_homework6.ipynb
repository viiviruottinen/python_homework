```python
# What do these two programs print? In simple terms, explain the difference between sorted(letters) and letters.sort().

# Program A
letters = list('gold')
result = sorted(letters)
print('letters is', letters, 'and result is', result)

# Prints the letters and alphabetically sorted letters separately.

# Program B
letters = list('gold')
result = letters.sort()
print('letters is', letters, 'and result is', result)

# Prints original letters in sorted form and doesn't print anything in result.
```

```python
# What do these two programs print? In simple terms, explain the difference between new = old and new = old[:].

# Program A
old = list('gold')
new = old      # simple assignment
new[0] = 'D'
print('new is', new, 'and old is', old)

# Old is assingned to new.

# Program B
old = list('gold')
new = old[:]   # assigning a slice
new[0] = 'D'
print('new is', new, 'and old is', old)

# Old is assingned to new as a new list by slicing and the old list stays as original.
```
