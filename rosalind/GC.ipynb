```python
dict_DNA = {}
key = ""

with open('rosalind_gc.txt', 'r') as f:
    for line in f.read().splitlines():
        if line.startswith('>'):
            key = line[1:]
            dict_DNA[key] = ""
        else:
            dict_DNA[key] = dict_DNA[key] + line

max_id = ""
max_gc = 0.0

for dna_id, dna in dict_DNA.items():
    gc_content = ((dna.count('G') + dna.count('C')) / len(dna)) * 100
    if gc_content > max_gc:
        max_gc = gc_content
        max_id = dna_id

print(max_id)
print(max_gc)
```
