```python
s = open('rosalind_revc.txt', 'r').read()[::-1]
complement = ''

for DNA in s:
    if DNA == 'C':
        complement = complement + 'G'
    if DNA == 'G':
        complement = complement + 'C'
    if DNA == 'T':
        complement = complement + 'A'
    if DNA == 'A':
        complement = complement + 'T'

print(complement)
```
