```python
# n = months
# m = months the rabbits live
# reproduce m-1 months

# example:

# F1 = 1 [1, 0, 0]
# F2 = 1 [0, 1, 0]
# F3 = 2 [1, 0, 1]
# F4 = 2 [1, 1, 0]
# F5 = 3 [1, 1, 1]
# F6 = 4 [2, 1, 1]


def mortal_rabbits(n, m):
    # create a list that tracks the number of rabbit pairs by age
    age = [1] + [0] * (m - 1)

    for i in range(1, n):
        # all rabbits that are at least one month old can reproduce and produce a single pair each month
        # count the newborn pairs
        newborns = sum(age[1:])
        # take all the numbers in the age list except the last one and add newborns to the beginning of the list
        age = [newborns] + age[:-1]

    return sum(age)

print(mortal_rabbits(84, 18))
```
