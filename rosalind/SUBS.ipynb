```python
with open('rosalind_subs.txt', 'r') as f:
    s, t = f.read().splitlines()

steps = len(s) - len(t) + 1
sub_length = len(t)
substrings = []

for i in range(steps):
    if s[i:i + sub_length] == t:
        substrings.append(i + 1)

print(*substrings)
```
